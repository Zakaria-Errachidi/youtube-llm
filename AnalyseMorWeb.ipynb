```python
import pandas as pd
import glob

```

```python
files = glob.glob("data*.csv")
df_list = []

for file in files:
    df = pd.read_csv(file, encoding="utf-8-sig")
    df_list.append(df)

# Fusionner tous les dataframes
df_total = pd.concat(df_list, ignore_index=True)

# Sauvegarder le dataset final
df_total.to_csv("dataset_total.csv", index=False)

print("Fusion terminée !")
print("Nombre total de lignes :", df_total.shape[0])
print("Nombre total de colonnes :", df_total.shape[1])
```
